```python
import pandas as pd
import numpy as np

# Set seed so the random data is reproducible
np.random.seed(42)

# Create DataFrame with 20 rows
df = pd.DataFrame({
    "delivery_time": np.random.randint(20, 61, 20),
    "order_value": np.random.randint(200, 1501, 20),
    "tip_amount": np.random.randint(0, 201, 20),
    "rating": np.round(np.random.uniform(2.5, 5.0, 20), 1),
    "delivery_distance": np.round(np.random.uniform(1, 15, 20), 2)
})

# Display the DataFrame
print("Food Delivery Orders:")
print(df)

# Calculate correlation matrix
correlation_matrix = df.corr()

# Display correlation matrix
print("\nCorrelation Matrix:")
print(correlation_matrix)
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f"
)

plt.title("Food Delivery Order Correlation Matrix")
plt.show()
```

```python
import pandas as pd
import numpy as np
import seaborn as sns
import matplotlib.pyplot as plt

# Set seed for reproducible data
np.random.seed(42)

# Create DataFrame with 20 rows
df = pd.DataFrame({
    "delivery_time": np.random.randint(20, 61, 20),
    "order_value": np.random.randint(200, 1501, 20),
    "tip_amount": np.random.randint(0, 201, 20),
    "rating": np.round(np.random.uniform(2.5, 5.0, 20), 1),
    "delivery_distance": np.round(np.random.uniform(1, 15, 20), 2)
})

# Calculate correlation matrix
correlation_matrix = df.corr()

# Create heatmap
plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation_matrix,
    annot=True,
    cmap="coolwarm",
    fmt=".2f",
    linewidths=0.5
)

plt.title("Food Delivery Data - Correlation Heatmap")
plt.show()
```

```python

# Load the Flipkart mobile dataset
df = pd.read_csv("flipkart_mobiles_2026.csv")

# Select at least 4 numeric columns
numeric_columns = [
    "price",
    "rating",
    "ram",
    "battery"
]

# Make sure the selected columns are numeric
df[numeric_columns] = df[numeric_columns].apply(
    pd.to_numeric,
    errors="coerce"
)

# Calculate correlation matrix
correlation_matrix = df[numeric_columns].corr()

# Display correlation matrix
print("Correlation Matrix:")
print(correlation_matrix)

# Create heatmap
plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    linewidths=0.5
)

plt.title("Flipkart Mobile Features Correlation Heatmap")
plt.tight_layout()
plt.show()
```
