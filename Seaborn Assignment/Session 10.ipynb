```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Sample Spotify streaming data
data = {
    "day": [
        "Mon", "Tue", "Wed", "Thu",
        "Fri", "Sat", "Sun"
    ],
    "songs_streamed": [
        45, 60, 52, 70, 85, 110, 95
    ]
}

df = pd.DataFrame(data)

# Apply poster context
sns.set(context="poster", style="whitegrid")

# Create barplot
plt.figure(figsize=(10, 6))

sns.barplot(
    data=df,
    x="day",
    y="songs_streamed"
)

plt.xlabel("Day")
plt.ylabel("Number of Songs Streamed")
plt.title("Daily Spotify Songs Streamed")

plt.show()
```

```python

# Sample Zomato-style food order data
data = {
    "order_id": range(1, 21),
    "cuisine": [
        "Indian", "Chinese", "Italian", "Indian", "Chinese",
        "Indian", "Italian", "Indian", "Chinese", "Italian",
        "Indian", "Chinese", "Indian", "Italian", "Chinese",
        "Indian", "Italian", "Chinese", "Indian", "Indian"
    ]
}

df = pd.DataFrame(data)

# Set Seaborn style
sns.set_theme(style="whitegrid")

# Create countplot
plt.figure(figsize=(8, 5))

sns.countplot(
    data=df,
    x="cuisine",
    palette="colorblind"
)

plt.xlabel("Cuisine Type")
plt.ylabel("Number of Food Orders")
plt.title("Food Orders Distribution by Cuisine Type")

plt.show()
```

```python

# Simulated daily step counts for 2 weeks
data = {
    "day": [
        "Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun",
        "Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"
    ],
    "steps": [
        6500, 7200, 8100, 7600, 9000, 10500, 9800,
        7000, 7800, 8500, 8200, 9400, 11000, 10200
    ]
}

df = pd.DataFrame(data)

# Apply notebook context and muted color palette
sns.set(
    context="notebook",
    palette="muted",
    style="whitegrid"
)

# Create lineplot
plt.figure(figsize=(10, 5))

sns.lineplot(
    data=df,
    x="day",
    y="steps",
    marker="o",
    linewidth=2
)

plt.xlabel("Day")
plt.ylabel("Daily Steps")
plt.title("Daily Step Counts - Two Weeks")

plt.tight_layout()
plt.show()
```

```python

# Existing plot
data = {
    "category": ["Indian", "Chinese", "Italian", "Mexican"],
    "orders": [120, 95, 80, 65]
}

# Changed: deep → colorblind
# Changed: talk → poster
sns.set(
    context="poster",
    palette="colorblind",
    style="whitegrid"
)

sns.barplot(
    x="category",
    y="orders",
    data=data
)

plt.title("Food Orders by Cuisine")
plt.xlabel("Cuisine Type")
plt.ylabel("Number of Orders")

plt.show()
```

```python

# Sample Flipkart sales data
data = {
    "category": [
        "Mobiles",
        "Laptops",
        "TVs",
        "Headphones",
        "Accessories"
    ],
    "sales": [
        850000,
        650000,
        480000,
        320000,
        250000
    ]
}

df = pd.DataFrame(data)

# Flipkart-style Seaborn theme
sns.set(
    context="talk",
    palette="deep",
    style="whitegrid",
    font_scale=1.0,
    rc={
        "axes.grid": True,
        "grid.linestyle": "--",
        "grid.alpha": 0.5
    }
)

# Barplot
plt.figure(figsize=(10, 6))

sns.barplot(
    data=df,
    x="category",
    y="sales"
)

plt.title("Flipkart Sales by Product Category", fontsize=20)
plt.xlabel("Product Category")
plt.ylabel("Sales (₹)")

plt.tight_layout()
plt.show()
```
