```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load the Flipkart CSV file
df = pd.read_csv("flipkart_products.csv")

# Check the column names
print(df.columns)

# Create jointplot
sns.jointplot(
    data=df,
    x="rating",
    y="price",
    kind="scatter"
)

plt.show()
```
