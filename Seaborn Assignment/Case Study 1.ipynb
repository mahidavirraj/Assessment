```python
import pandas as pd

# Load Bank Marketing dataset
bank_df = pd.read_csv("bank.csv")

# Display first 10 rows
print("First 10 rows:")
display(bank_df.head(10))

# Display DataFrame information
print("\nDataFrame Information:")
bank_df.info()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# Set Seaborn style
sns.set_theme(style="whitegrid")

# Create countplot
plt.figure(figsize=(8, 5))

sns.countplot(
    data=bank_df,
    x="y"
)

plt.title("Term Deposit Subscription")
plt.xlabel("Subscribed to Term Deposit")
plt.ylabel("Number of Customers")

plt.tight_layout()
plt.show()
```

```python
# Create boxplot
sns.boxplot(x='y', y='age', data=bank_df)

# Add labels and title
plt.xlabel('Term Deposit Subscription')
plt.ylabel('Customer Age')
plt.title('Customer Age vs Term Deposit Subscription')

# Display the plot
plt.show()
```

```python
# Calculate percentage of customers who subscribed
education_subscription = (
    bank_df.groupby('education')['y']
      .apply(lambda x: (x == 'yes').mean() * 100)
      .reset_index(name='subscription_percentage')
)

# Create bar plot
sns.barplot(
    data=education_subscription,
    x='education',
    y='subscription_percentage'
)

plt.xlabel('Education Level')
plt.ylabel('Subscription Percentage (%)')
plt.title('Term Deposit Subscription by Education Level')
plt.xticks(rotation=45)
plt.show()
```

```python
# Calculate subscription percentage for each job category
job_subscription = (
    bank_df.groupby('job')['y']
      .apply(lambda x: (x == 'yes').mean() * 100)
      .reset_index(name='subscription_percentage')
)

# Create bar plot
plt.figure(figsize=(12, 6))

sns.barplot(
    data=job_subscription,
    x='job',
    y='subscription_percentage'
)

plt.xlabel('Job Type')
plt.ylabel('Subscription Percentage (%)')
plt.title('Term Deposit Subscription Rate by Job Type')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```
