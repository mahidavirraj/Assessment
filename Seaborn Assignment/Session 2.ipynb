```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Sample Swiggy orders data
data = {
    'order_id': [101, 102, 103, 104, 105, 106, 107, 108, 109, 110],
    'payment_method': [
        'UPI', 'Card', 'UPI', 'Cash', 'Card',
        'UPI', 'Card', 'Cash', 'UPI', 'Card'
    ]
}

df = pd.DataFrame(data)

# Set theme
sns.set_theme(style='whitegrid')

# Count orders by payment method
sns.countplot(data=df, x='payment_method')

plt.xlabel('Payment Method')
plt.ylabel('Number of Orders')
plt.title('Number of Orders by Payment Method')
plt.show()
```

```python
# Sample Zomato-style dataset
data = {
    'food_category': [
        'Pizza', 'Pizza', 'Pizza',
        'Burger', 'Burger', 'Burger',
        'Biryani', 'Biryani', 'Biryani',
        'South Indian', 'South Indian', 'South Indian'
    ],
    'delivery_time': [
        30, 35, 32,
        25, 28, 30,
        40, 45, 42,
        35, 38, 36
    ]
}

df = pd.DataFrame(data)

# Set Seaborn theme
sns.set_theme(style='whitegrid')

# Create barplot showing average delivery time
sns.barplot(
    data=df,
    x='food_category',
    y='delivery_time',
    estimator='mean'
)

plt.xlabel('Food Category')
plt.ylabel('Average Delivery Time (minutes)')
plt.title('Average Delivery Time by Food Category')
plt.show()
```

```python
# Sample Zomato-style dataset
data = {
    'food_category': [
        'Pizza', 'Pizza', 'Pizza',
        'Burger', 'Burger', 'Burger',
        'Biryani', 'Biryani', 'Biryani',
        'South Indian', 'South Indian', 'South Indian'
    ],
    'delivery_time': [
        30, 35, 32,
        25, 28, 30,
        40, 45, 42,
        35, 38, 36
    ]
}

df = pd.DataFrame(data)

sns.set_theme(style='whitegrid')

# Barplot using sum
sns.barplot(
    data=df,
    x='food_category',
    y='delivery_time',
    estimator='sum'
)

plt.xlabel('Food Category')
plt.ylabel('Total Delivery Time (minutes)')
plt.title('Total Delivery Time by Food Category')
plt.show()
```

```python
# Sample BookMyShow-style dataset
data = {
    'movie': [
        'Movie A', 'Movie B', 'Movie C', 'Movie D',
        'Movie E', 'Movie F', 'Movie G', 'Movie H',
        'Movie I', 'Movie J'
    ],
    'genre': [
        'Action', 'Comedy', 'Action', 'Drama',
        'Comedy', 'Horror', 'Action', 'Drama',
        'Comedy', 'Horror'
    ]
}

df = pd.DataFrame(data)

sns.set_theme(style='whitegrid')

# Count movies by genre
sns.countplot(data=df, x='genre')

plt.xlabel('Movie Genre')
plt.ylabel('Number of Movies')
plt.title('Number of Movies by Genre')
plt.show()
```

```python
sns.barplot(
    data=df,
    x='genre',
    y='movie',
    estimator='count',
    ci=None
)

plt.xlabel('Movie Genre')
plt.ylabel('Number of Movies')
plt.title('Number of Movies by Genre')
plt.show()
```
