```python
import seaborn as sns
import matplotlib.pyplot as plt

# Load the tips dataset
tips = sns.load_dataset("tips")

# Create pairplot for numerical columns
sns.pairplot(tips)

plt.show()
```

```python
sns.pairplot(
    tips[['total_bill', 'tip', 'size']]
)

plt.show()
```

```python

# Load the tips dataset
tips = sns.load_dataset("tips")

# Create pairplot with sex as hue
sns.pairplot(
    tips,
    hue="sex"
)

plt.show()
```

```python
sns.pairplot(tips, hue="sex", palette="Set1")
plt.show()
```

```python

# Load the tips dataset
tips = sns.load_dataset("tips")

# Create pairplot
sns.pairplot(
    tips,
    hue="smoker",
    diag_kind="kde"
)

plt.show()
```

```python

# Load the tips dataset
tips = sns.load_dataset("tips")

# Pairplot using vars to select only two numerical columns
sns.pairplot(
    tips,
    vars=["total_bill", "tip"],
    hue="time"
)

plt.show()
```
