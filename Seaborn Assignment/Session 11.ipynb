```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Sample food order data
data = {
    "food_category": [
        "Pizza", "Pizza", "Pizza",
        "Burger", "Burger", "Burger", "Burger",
        "Biryani", "Biryani",
        "Momos", "Momos", "Momos", "Momos",
        "Ice Cream", "Ice Cream"
    ]
}

df = pd.DataFrame(data)

# Seaborn theme
sns.set_theme(style="whitegrid")

# Create barplot
plt.figure(figsize=(9, 6))

sns.countplot(
    data=df,
    x="food_category"
)

plt.title("Number of Orders by Food Category")
plt.xlabel("Food Category")
plt.ylabel("Number of Orders")

plt.tight_layout()

# Save as PNG with 300 DPI
plt.savefig(
    "food_category_orders.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```

```python
# Daily step counts for one week
data = {
    "day": ["Mon", "Tue", "Wed", "Thu", "Fri", "Sat", "Sun"],
    "steps": [6500, 7200, 8100, 7600, 9000, 10500, 9800]
}

df = pd.DataFrame(data)

# Seaborn theme
sns.set_theme(style="whitegrid")

# Create line plot
plt.figure(figsize=(10, 5))

sns.lineplot(
    data=df,
    x="day",
    y="steps",
    marker="o",
    linewidth=2
)

plt.title("Daily Step Counts - One Week")
plt.xlabel("Day")
plt.ylabel("Number of Steps")

plt.tight_layout()

# Export as SVG
plt.savefig(
    "weekly_step_counts.svg",
    format="svg",
    bbox_inches="tight"
)

plt.show()
```

```python
# Sample sales data
data = {
    "category": ["Mobiles", "Laptops", "TVs", "Headphones", "Accessories"],
    "sales": [850000, 650000, 480000, 320000, 250000]
}

df = pd.DataFrame(data)

# Customize Seaborn theme
sns.set_theme(
    style="darkgrid",
    palette="muted"
)

# Create chart
plt.figure(figsize=(10, 6))

sns.barplot(
    data=df,
    x="category",
    y="sales"
)

plt.title("Sales by Product Category")
plt.xlabel("Product Category")
plt.ylabel("Sales (₹)")

plt.tight_layout()

# Save as high-resolution PNG
plt.savefig(
    "sales_chart_1200dpi.png",
    dpi=1200,
    format="png",
    bbox_inches="tight"
)

plt.show()
```

```python
# Reusable presentation theme
def apply_presentation_theme():
    sns.set(
        style="whitegrid",
        context="talk",
        palette="deep",
        font="DejaVu Sans",
        font_scale=1.0
    )

    plt.rcParams["figure.figsize"] = (10, 6)
    plt.rcParams["axes.titlesize"] = 20
    plt.rcParams["axes.labelsize"] = 14


# Apply the theme
apply_presentation_theme()


# Top 5 most-used apps
data = {
    "app": ["YouTube", "Instagram", "WhatsApp", "Spotify", "Facebook"],
    "hours": [25, 20, 18, 12, 8]
}

df = pd.DataFrame(data)


# Create pie chart
plt.figure(figsize=(9, 7))

plt.pie(
    df["hours"],
    labels=df["app"],
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Top 5 Most-Used Apps")
plt.tight_layout()


# Export chart
plt.savefig(
    "top_5_most_used_apps.png",
    dpi=300,
    bbox_inches="tight"
)

plt.show()
```
