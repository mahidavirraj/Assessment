```python
import seaborn as sns
import matplotlib.pyplot as plt

print("Seaborn version:", sns.__version__)
print("Matplotlib version:", plt.matplotlib.__version__)
```

```python
days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']
temps = [32, 34, 33, 35, 36, 37, 34]

sns.lineplot(x=days, y=temps, marker='o')

plt.xlabel('Day')
plt.ylabel('Temperature (°C)')
plt.title('Weekly Temperature Trend')

plt.show()
```

```python
days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']
temps = [32, 34, 33, 35, 36, 37, 34]

sns.relplot(
    x=days,
    y=temps,
    kind='line',
    marker='o'
)

plt.xlabel('Day')
plt.ylabel('Temperature (°C)')
plt.title('Weekly Temperature Trend')
plt.show()
```

```python
days = ['Mon', 'Tue', 'Wed', 'Thu', 'Fri', 'Sat', 'Sun']
temps = [32, 34, 33, 35, 36, 37, 34]

# 1. Darkgrid theme
sns.set_theme(style='darkgrid')

plt.figure(figsize=(8, 5))
sns.lineplot(x=days, y=temps, marker='o')
plt.xlabel('Day')
plt.ylabel('Temperature (°C)')
plt.title('Weekly Temperature Trend - Darkgrid')
plt.savefig('temperature_darkgrid.png', dpi=300, bbox_inches='tight')
plt.show()


# 2. Whitegrid theme
sns.set_theme(style='whitegrid')

plt.figure(figsize=(8, 5))
sns.lineplot(x=days, y=temps, marker='o')
plt.xlabel('Day')
plt.ylabel('Temperature (°C)')
plt.title('Weekly Temperature Trend - Whitegrid')
plt.savefig('temperature_whitegrid.png', dpi=300, bbox_inches='tight')
plt.show()


# 3. Ticks theme
sns.set_theme(style='ticks')

plt.figure(figsize=(8, 5))
sns.lineplot(x=days, y=temps, marker='o')
plt.xlabel('Day')
plt.ylabel('Temperature (°C)')
plt.title('Weekly Temperature Trend - Ticks')
plt.savefig('temperature_ticks.png', dpi=300, bbox_inches='tight')
plt.show()
```
