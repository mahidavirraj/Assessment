```python
# Sample monthly Zomato spending amounts (₹)
monthly_spending = [
    250, 450, 600, 750, 900,
    1100, 1250, 1350, 1500, 1650,
    1800, 1950, 2100, 2250, 2400,
    2600, 2800, 3000, 3200, 3500,
    3800, 4000, 4300, 4600, 5000,
    5500, 6000, 6500, 7000, 8000
]

# Set theme
sns.set_theme(style="whitegrid")

# Create a single figure and axes
fig, ax = plt.subplots(figsize=(9, 5))

# Histogram
sns.histplot(
    monthly_spending,
    bins=10,
    stat="density",
    kde=False,
    ax=ax
)

# KDE curve on the same axes
sns.kdeplot(
    monthly_spending,
    ax=ax,
    linewidth=2
)

plt.xlabel("Monthly Spending (₹)")
plt.ylabel("Density")
plt.title("Distribution of Monthly Zomato Spending")

plt.show()
```
