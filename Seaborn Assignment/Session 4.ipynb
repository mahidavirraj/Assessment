```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Load CSV file
df = pd.read_csv("flipkart_products.csv")

# Display first few rows
print(df.head())

# Set Seaborn theme
sns.set_theme(style="whitegrid")

# Create boxplot
plt.figure(figsize=(8, 5))
sns.boxplot(data=df, y="price")

plt.xlabel("Product")
plt.ylabel("Price (₹)")
plt.title("Flipkart Product Price Distribution")

plt.show()
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Sample Zomato data
data = {
    'category': [
        'Fast Food', 'Fast Food', 'Fast Food', 'Fast Food', 'Fast Food',
        'Desserts', 'Desserts', 'Desserts', 'Desserts', 'Desserts'
    ],
    'order_amount': [
        250, 300, 450, 200, 350,
        150, 220, 300, 180, 400
    ]
}

df = pd.DataFrame(data)

# Set theme
sns.set_theme(style="whitegrid")

# Violin plot
plt.figure(figsize=(8, 5))
sns.violinplot(
    data=df,
    x='category',
    y='order_amount'
)

plt.xlabel('Restaurant Category')
plt.ylabel('Order Amount (₹)')
plt.title('Zomato Order Amounts: Fast Food vs Desserts')
plt.show()
```

```python
import seaborn as sns
import matplotlib.pyplot as plt

# IPL match ticket prices in ₹
ticket_prices = [
    500, 750, 1000, 1200, 1500,
    1800, 2000, 2200, 2500, 3000,
    1200, 1500, 1800, 2000, 2500,
    3500, 4000, 4500, 5000, 15000
]

# Create boxplot
sns.set_theme(style="whitegrid")

plt.figure(figsize=(8, 5))
sns.boxplot(y=ticket_prices)

plt.ylabel("Ticket Price (₹)")
plt.title("IPL Match Ticket Price Distribution")
plt.show()
```

```python
import numpy as np

Q1 = np.percentile(ticket_prices, 25)
Q3 = np.percentile(ticket_prices, 75)

IQR = Q3 - Q1

lower_limit = Q1 - 1.5 * IQR
upper_limit = Q3 + 1.5 * IQR

outliers = [
    price for price in ticket_prices
    if price < lower_limit or price > upper_limit
]

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)
print("Outliers:", outliers)
```

```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt

# Sample daily step counts for 30 days
steps = [
    4200, 5100, 6500, 7200, 8000, 8500, 9000, 9500, 10000, 10500,
    6200, 7000, 7600, 8200, 8800, 9300, 9800, 10200, 11000, 11500,
    4500, 5600, 6800, 7500, 8100, 8700, 9200, 9700, 10800, 12000
]

df = pd.DataFrame({
    'daily_steps': steps
})

# Set theme
sns.set_theme(style="whitegrid")

# Create violin plot
plt.figure(figsize=(8, 5))

sns.violinplot(
    data=df,
    y='daily_steps'
)

plt.ylabel("Daily Steps")
plt.title("Distribution of Daily Step Counts")
plt.show()
```
